```python
# 3장 1강 : 시각화의 목적과 차트 선택 프레임 워크

# 오늘의 코딩 테스트 Greedy - 각 단계에서 현재 상황에서 가장 좋은 선택을 반복하는 문제 해결 방식

# 가능한 한 많은 모델 실행 > 유리한 선택 > greedy 사용
# limit을 초과 할 수 없다.  1 <= limit <= 100
# 하나의 모델은 한번만 실행 > 반복문 > for문 사용
# 리스트 정렬 후 for 조건문 사용해서 반환하고  함수에 greedy 사용


# 많이 실행하려면 어떤 모델부터 골라야 유리할까 > 많이 실행하려면 메모리를 적게 사용하는 모델부터 선택하는게 유리하다

def solution (models,limit):
    models = sorted(models) #정렬 > 작은 모델부터 봐야 최대한 많이 넣을 수 있음

    used_memory = 0 # 지금까지 gpu 메모리를 얼마나 사용했는가
    count = 0 # 몇개를 실행할 건지 카운트 할 수 있는 변수를 0

    for memory in models: # 오름차순으로 정렬 된 model 순회
        if used_memory + memory <= limit:
            used_memory += memory
            count += 1
        else:
            break

    return count

print(solution([4, 2, 5, 1], 7))
print(solution([6, 3, 2, 4], 5))
print(solution([5, 6, 7], 4))
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

online = pd.read_csv('./ecommerce_orders_online.csv')
offline = pd.read_csv('./ecommerce_orders_offline.csv')

online['channel'] = 'Online'
offline['channel'] = 'Offline'

df = pd.concat(
    [online, offline],
    ignore_index=True
)

sns.barplot(
    data=df,
    x='product_category',
    y='revenue',
    hue='channel',
    errorbar=('ci', 95)
)

plt.xlabel('Product Category')
plt.ylabel('Average Revenue')
plt.title('Average Revenue by Category and Channel')

plt.show()
```
