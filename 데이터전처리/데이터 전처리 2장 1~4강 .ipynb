```python
# 2장 3강 
import pandas as pd

df = pd.DataFrame({
    "주문금액": [20000, 22000, 25000, 27000, 30000, 32000, 35000, 38000, 40000, 300000]
})

Q1 = df["주문금액"].quantile(0.25) # 하위 기준 25%
Q3 = df["주문금액"].quantile(0.75) # 상위 기준 25%

IQR = Q3 - Q1 # 특정 숫자
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 = 1.5 * IQR

# 이상치 탐지
outliers = df[(df['값'] < lower_bound) | (df['값'] > upper_bound)]
print(f'IQR 범위: {lower_bound:.2f} ~ {upper_bound:.2f}')
print(f'이상치:\n{outliers}')
```

```python
import pandas as pd

df = pd.read_csv(
    './ecommerce_orders_online.csv'
)

Q1 = df['revenue'].quantile(0.25)
Q3 = df['revenue'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 = 1.5 * IQR

outliers = df[(df['revenue'] < lower_bound) |  (df['revenue'] > upper_bound)]


print('하한값:', lower_bound)
print('상한값:', upper_bound)
print('이상치 개수:', len(outliers))


print(
    outliers[
        ['order_id', 'product_category', 'revenue']
    ].head()
)
```

```python
import pandas as pd

df = pd.read_csv(
    './customer_order_summary.csv'
)

Q1 = df["total_spent"].quantile(0.25)
Q3 = df['total_spent'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR


outliers = df[
    (df['total_spent'] < lower_bound) |
    (df['total_spent'] > upper_bound)
]

print('이상치 개수:', len(outliers))

print(
    outliers[
        ['customer_id', 'total_orders', 'total_spent']
    ].head())
```
