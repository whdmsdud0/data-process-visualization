```python
#오늘의 코딩 테스트 정렬 sorting

def solution(priorities):
    sorted_priorities = sorted(priorities, reverse=True)

    return sorted_priorities

  
print(solution([3, 1, 5, 2, 4]))
print(solution([2, 5, 2, 4, 5]))
print(solution([7]))
```

```python
# join(): 인덱스 기반 결합
# pandas의 join은 반드시 인덱스가 있어야 함
# join() merge() 차이점: 둘 다 두 개의 dataframe을 합치는 것인데, 어떤 기준으로 합치느냐가 핵심 차이
# join은 인덱스를 기준으로 합치는 데 편함
# merge는 특정 컬럼을 기준으로 합치는데 편함

# 연습문제 8.
# 가장 먼저 해야 하는 것 csv 파일 읽어오기


import pandas as pd
customer_master = pd.read_csv(
    './customer_master.csv'
)
customer_order_summary = pd.read_csv(
    './customer_order_summary.csv'
)

# join을 사용해서 두 개의 데이터프레임을 결합 > 인덱스 필요
customer_master_idx = customer_master.set_index('customer_id')
customer_order_summary_idx = customer_order_summary.set_index('customer_id')

df_customer_view = customer_master_idx.join(  # 고객 마스터의 데이터가 모두 유지되도록 left join
    customer_order_summary_idx,
    how='left'
)

df_customer_view.head()
```

```python
import pandas as pd

customer_master = pd.read_csv('./customer_master.csv')
customer_order_summary = pd.read_csv('./customer_order_summary.csv')

df_customer = pd.merge(
    customer_master,
    customer_order_summary,
    on='customer_id',
    how='inner'
)

total_spent = (
    df_customer.groupby('membership_tier')['total_spent']
    .sum()
    .reset_index()
)

avg_spent = (
    df_customer.groupby('membership_tier')['total_spent']
    .mean()
    .reset_index()
)

print(total_spent)
print(avg_spent)
```
